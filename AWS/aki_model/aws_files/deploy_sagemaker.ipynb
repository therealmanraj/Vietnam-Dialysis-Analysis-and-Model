```python
import sagemaker
from sagemaker.sklearn.model import SKLearnModel

# ── CONFIG ──────────────────────────────────────────
AWS_REGION   = "ca-central-1"
S3_BUCKET    = "my-aki-model-bucket"
S3_MODEL_KEY = "models/xgb_aki_model.tar.gz"               # <--- tar.gz, not .joblib
ROLE_ARN     = "arn:aws:iam::222634404112:role/SageMakerExecutionRole-ca"  
ENDPOINT_NAME = "aki-xgb-endpoint"

# ── INIT SESION ─────────────────────────────────────
sess = sagemaker.Session()
role = ROLE_ARN

# ── DEFINE THE MODEL USING THE TAR.GZ IN S3 ─────────
sk_model = SKLearnModel(
    model_data=f"s3://{S3_BUCKET}/{S3_MODEL_KEY}",
    role=role,
    entry_point="inference.py",
    framework_version="0.23-1",  # or whichever is compatible
    py_version="py3",
    sagemaker_session=sess
)

# ── DEPLOY ──────────────────────────────────────────
predictor = sk_model.deploy(
    initial_instance_count=1,
    instance_type="ml.m5.large",
    endpoint_name=ENDPOINT_NAME
)

print(f"✅ Deployed endpoint: {ENDPOINT_NAME}")
```
